```python
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from requests.exceptions import HTTPError, Timeout, RequestException
import time

def test_with_beautiful_soup(url):
    try:
        # Request the page using requests
        response = requests.get(url, timeout=10)
        
        # If the request is successful, parse the page with BeautifulSoup
        if response.status_code == 200:
            print("BeautifulSoup Test: Successfully connected and fetched the page.")
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Try to find the title of the page as a simple test
            title = soup.title.string if soup.title else "No title found"
            print(f"BeautifulSoup Test: Page title: {title}")
        else:
            print(f"BeautifulSoup Test: Failed to fetch the page. Status Code: {response.status_code}")
    
    except (HTTPError, Timeout, RequestException) as e:
        print(f"BeautifulSoup Test: Error occurred - {e}")

def test_with_selenium(url):
    try:
        # Setup Selenium WebDriver (using Chrome here)
        chrome_options = Options()
        chrome_options.add_argument("--headless")  # Run in headless mode (without opening browser window)
        
        # Specify the path to your ChromeDriver
        driver_path = r"C:\chromedriver\chromedriver-win64\chromedriver.exe"  # Replace with your chromedriver path
        driver_service = Service(driver_path)
        
        # Start the WebDriver
        driver = webdriver.Chrome(service=driver_service, options=chrome_options)
        
        # Fetch the page
        driver.get(url)
        
        # Allow time for the page to load (adjust as necessary)
        time.sleep(3)
        
        # Test if we can find the title of the page
        title = driver.title
        print(f"Selenium Test: Page title: {title}")
        
        # Close the driver
        driver.quit()
    
    except Exception as e:
        print(f"Selenium Test: Error occurred - {e}")

if __name__ == "__main__":
    # URL to scrape
    test_url = "https://www.pro-football-reference.com"
    
    # Test with BeautifulSoup
    test_with_beautiful_soup(test_url)
    
    # Test with Selenium
    test_with_selenium(test_url)
```
